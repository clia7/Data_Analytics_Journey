```python
# Convert into pandas daytime
fe['date'] = pd.to_datetime({
    'year': df['year'],
    'month': df['month'],
    'day': df['day_of_month'] 
})
```

```python
# Convert time
time_convert = ['crs_dep_time', 'dep_time', 'wheels_off',
       'wheels_on', 'crs_arr_time', 'arr_time']

for cols in time_convert:
    fe[cols] = pd.to_datetime(
        fe[cols].astype('Int64').astype(str).str.zfill(4), 
        format='%H%M',
        errors='coerce'
    )
```

```python
fe['crs_datetime'] = pd.to_datetime(
    fe['date'].dt.strftime('%Y-%m-%d') + ' ' +
    fe['crs_dep_time'].dt.strftime('%H:%M:%S')
)

fe['dep_datetime'] = pd.to_datetime(
    fe['date'].dt.strftime('%Y-%m-%d') + ' ' +
    fe['dep_time'].dt.strftime('%H:%M:%S')
)

fe['arr_date'] = fe['date'].copy()
fe.loc[fe['arr_time'] < fe['dep_time'], 'arr_date'] += pd.Timedelta(days=1)

fe['crs_arr_date'] = fe['date'].copy()
fe.loc[fe['crs_arr_time'] < fe['crs_dep_time'], 'crs_arr_date'] += pd.Timedelta(days=1)

fe['crs_arr_datetime'] = pd.to_datetime(
    fe['crs_arr_date'].dt.strftime('%Y-%m-%d') + ' ' +
    fe['crs_arr_time'].dt.strftime('%H:%M:%S')
)

fe['arr_datetime'] = pd.to_datetime(
    fe['arr_date'].dt.strftime('%Y-%m-%d') + ' ' +
    fe['arr_time'].dt.strftime('%H:%M:%S')
)
```

```python
# Missing Values 
missing_values = pd.DataFrame({
    'Missing': fe.isna().sum(),
    'Percentage': fe.isna().mean() * 100
})

missing_values = missing_values[missing_values['Missing'] > 0]
missing_values.sort_values(by='Percentage', ascending=False)
```

**Missing Value Handling – Initial Steps**

- **`cancellation_code`:** Dropped due to a high percentage of missing values (98.64%). The column is only populated for cancelled flights and is not required for the prediction task.
- **Rows with a single missing value:** Removed because the number of affected records is negligible compared to the overall dataset size.

```python
# Drop Cancellation code
fe.drop('cancellation_code', axis='columns', inplace=True)
```

```python
# Drop the rows with 1 missing value
fe.dropna(subset=['crs_dep_time', 'crs_elapsed_time', 'crs_datetime', 'op_carrier_fl_num'], inplace=True)
```

```python
fe.drop(columns=['wheels_off',
                 'wheels_on', 'taxi_out', 'taxi_in', 'actual_elapsed_time',
                 'air_time', 'arr_time', 'arr_datetime', 'dep_time', 'dep_time', 'dep_datetime', 'dep_delay'], inplace=True)
```

```python
# Missing Values 
missing_values = pd.DataFrame({
    'Missing': fe.isna().sum(),
    'Percentage': fe.isna().mean() * 100
})

missing_values = missing_values[missing_values['Missing'] > 0]
missing_values.sort_values(by='Percentage', ascending=False)
```

**Missing Value Analysis – Remaining Columns**

The remaining missing values are mainly found in actual flight information:

- `arr_delay`: Retained as the prediction target.
- `air_time`: Excluded because actual flight duration is unknown before departure.
- `arr_time`, `arr_datetime`, `dep_time`, `dep_datetime`: Actual timestamps, not available before departure.
- `dep_delay`: Excluded due to data leakage, as the actual departure delay is unknown before departure.

**Decision:** These columns will be excluded during final feature selection. `crs_elapsed_time` (scheduled flight duration) will be retained as a predictive feature.

```python
fe.groupby('cancelled').agg(
    total_flights=('arr_delay', 'size'),
    missing_arr_delay=('arr_delay', lambda x: x.isna().sum())
)
```
