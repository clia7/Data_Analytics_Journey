```python
# Import libraries
import pandas as pd
import numpy as np
import seaborn as sns

# Safe dataloading for both datasets
file_with_dup = 'hotel_booking_with_dup.csv'
file_no_dup = 'hotel_booking_no_dup.csv'
try:
    df_with_dup = pd.read_csv(file_with_dup)
    print(f'File: {file_with_dup} load successfully')
except FileNotFoundError:
    print(f'File not found - {file_with_dup}')
try:
    df_no_dup = pd.read_csv(file_no_dup)
    print(f'File: {file_no_dup} load successfully')
except FileNotFoundError:
    print(f'File not found - {file_no_dup}')
```

**Note:** As the working process is now with two datasets, I will start implementing functions to reduce typing errors.

```python
# Starting with the general inspection for both datasets and implement a function to reduce error

def data_inspection(df, name):
    """
    Inspection function to safely analyze dataset structures,
    missing values, and data types without code duplication
    """
    print("=" * 60)
    print(f'Dataset inspection for: {name}')
    print("=" * 60)
    print('Dataset Shape:')
    print(f'Rows: {df.shape[0]} -- Columns {df.shape[1]}')
    print('\n')
    print('Missing Values')
    missing = df.isna().sum()
    print(missing[missing > 0] if missing.sum() > 0 else 'No Missing Values found')
    print('\n')
    print('Datatypes & Info:')
    df.info()
    print('\n')
    print('Header & First Rows')
    display(df.head())
```

```python
# Check the datasest without duplicates with function
data_inspection(df_no_dup, 'Dataset without Duplicates')
```

```python
# Check the datasest with duplicates with function
data_inspection(df_with_dup, 'Dataset with Duplicates')
```

```python
# Implement function to calculate the total number of pax by accumulate adults + children + babies
def calculate_total_pax(df):
    """
    Summarise the total number of pax by accumulate column adults + children + babies
    """
    df['total_pax'] = df['adults'] + df['children'] + df['babies']
    df['total_pax'] = df['total_pax'].astype(int)
    return df
```

```python
# Call funtion to calculate for both datasets
df_no_dup = calculate_total_pax(df_no_dup)
df_with_dup = calculate_total_pax(df_with_dup)
```

```python
# Implement function to drop the adr outliers of minus and > 5000$
def remove_adr_outliers(df):
    """
    Remove the extreme outliers in the dataset with minus adr and adr over 5000
    """
    df = df[(df['adr'] >= 0) & (df['adr'] < 5000)].copy()
    return df
```

```python
# Call function for both datasets to remove the extreme outliers
df_no_dup = remove_adr_outliers(df_no_dup)
df_with_dup = remove_adr_outliers(df_with_dup)
```

```python
# Implement function to check for changes between reserved room type and assigned room type
def check_roomtype_changes(df):
    """
    Check for changes between reserved room type and assigned room type
    """
    df['room_type_changes'] = (df['reserved_room_type'] != df['assigned_room_type'] ).astype(int)
    return df
```

```python
# Call function for roomtype changes
df_no_dup = check_roomtype_changes(df_no_dup)
df_with_dup = check_roomtype_changes(df_with_dup)
```

```python
# Implement function to change agent and company into binary
def create_agent_company_binary(df):
    """
    Create new column has agent and has company and convert the numbers into binary
    """
    df['has_agent'] = (df['agent'] != 0).astype(int)
    df['has_company'] = (df['company'] != 0).astype(int)
    df = df.drop(columns=['company', 'agent'])
    return df
```

```python
# Call function to change agent and company
df_no_dup = create_agent_company_binary(df_no_dup)
df_with_dup = create_agent_company_binary(df_with_dup)
```

```python
# Implement function to seperate the countries into domestic, european and international
def create_country_regions(df):
    """
    Changes the countries into three main categories: domestic for portugal, european for all european countries and international for all other countries
    """
    europe_codes = [
        'GBR', 'FRA', 'ESP', 'DEU', 'ITA', 'IRL', 'BEL', 'NLD', 
        'CHE', 'AUT', 'SWE', 'POL', 'ISR', 'NOR']
    
    df['country_region'] = 'International'
    df.loc[df['country'].isin(europe_codes), 'country_region'] = 'Europe'
    df.loc[df['country'] == 'PRT', 'country_region'] = 'Domestic'

    df = pd.get_dummies(df, columns=['country_region'], dtype=int)

    df = df.drop(columns=['country'])
    return df
```

```python
# Call function to seperate the countries into regions
df_no_dup = create_country_regions(df_no_dup)
df_with_dup = create_country_regions(df_with_dup)
```

```python
# Implement function to encode object columns with pd.get_dummies
def encode_categorial_cols(df):
    """
    Encode the remaining object (text) columns into binary features with One-Hot-Encode
    """
    columns_to_encode = ['hotel', 'meal', 'market_segment', 'distribution_channel', 'deposit_type', 'customer_type']
    df = pd.get_dummies(df, columns=columns_to_encode, dtype=int)
    return df
```

```python
# Call function to encode
df_no_dup = encode_categorial_cols(df_no_dup)
df_with_dup = encode_categorial_cols(df_with_dup)
```
