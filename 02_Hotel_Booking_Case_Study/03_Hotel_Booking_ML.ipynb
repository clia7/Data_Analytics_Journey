# Hotel Booking Demand Analysis
## Machine Learning
* **Goal:** Predict Booking Cancellations

```python
# Import relevant libraries
import pandas as pd
import numpy as np 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
```

```python
# Loading ML Dataset
file_path = 'hotel_booking_ml.csv'
try:
    df = pd.read_csv(file_path)
    print('Data loaded successfully')
except FileNotFoundError:
    print(f' Data not found: Check {file_path}')
```

```python
# Inspection of Dataset
df.head()
```

```python
# Splitting the Dataset into X the features and y the target
X = df.drop(columns='is_canceled')
y = df['is_canceled']

# Creating the Train, test variables
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Checking shape for train and test set
print(f'Train Set: {X_train.shape} -- Train Set: {y_train.shape[0]}')
print(f'Test Set: {X_test.shape} -- Test Set: {y_test.shape[0]}')
```

```python
# Prepare data with StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Implement LogisticRegressionModel
lr_model = LogisticRegression(max_iter=1000)

# Train the model
lr_model.fit(X_train_scaled, y_train)
```

```python
# Make prediction with the trained model
Lr_pred = lr_model.predict(X_test_scaled)

# Checking Accuracy Score
lr_accuracy = accuracy_score(y_test, Lr_pred)

# Output of Accuracy Score
print(f'Score: {lr_accuracy*100:.2f}%')
```

```python
# Checking the coefficience
lr_coeff = pd.DataFrame({'Feature': X.columns, 'Ranking': lr_model.coef_[0]})
lr_coeff = lr_coeff.sort_values(by='Ranking', ascending=False)
print(lr_coeff)
```

```python
# Implement RandomForestClassifier
rf_model = RandomForestClassifier(random_state=42)

# Train the model
rf_model.fit(X_train, y_train)
```

```python
# Make prediction with the trained model
rf_pred = rf_model.predict(X_test)

# Accuracy Score of the model
rf_accuracy = accuracy_score(y_test, rf_pred)

# Output of Accuracy Score
print(f'Accuracy Score: {rf_accuracy*100:.2f}%')
```

```python
# Checking the feature importance
rf_features = pd.DataFrame({'Features': X.columns, 'Importance': rf_model.feature_importances_})
rf_features = rf_features.sort_values(by='Features', ascending=False)
print(rf_features)
```

```python
# Import classification report 
from sklearn.metrics import classification_report

# Output of Classification Report of LR Model
print(classification_report(y_test, Lr_pred))
```

```python
# Checking Classification Report for RF Model
print(classification_report(y_test, rf_pred))
```
