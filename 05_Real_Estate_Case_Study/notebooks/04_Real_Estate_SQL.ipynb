```python
# Import relevant libraries

import pandas as pd 
import sqlite3

# Import CSV
df = pd.read_csv('../data/world_real_estate_data(147k).csv')

# Create SQLite database and table
with sqlite3.connect('../data/real_estate.db') as conn:
    df.to_sql(
        name='real_estate', 
        con=conn, 
        index=False, 
        if_exists='replace'
)

print('Database successfully created and loaded')
```

```python
%%sql

SELECT *
FROM real_estate
LIMIT 5;
```

```python
%%sql

SELECT
    country, 
    location, 
    apartment_total_area, 
    price_in_USD
FROM real_estate
LIMIT 10;
```

```python
%%sql

SELECT DISTINCT
    country
FROM real_estate;
```

```python
%%sql

SELECT
    title AS property_title, 
    country AS property_country, 
    price_in_USD AS price_usd
FROM real_estate
LIMIT 10;
```

```python
%%sql

SELECT
    title, 
    location, 
    price_in_USD
FROM real_estate
WHERE country = 'Turkey'
LIMIT 10;
```

```python
%%sql

SELECT
    title, 
    country, 
    price_in_USD
FROM real_estate
WHERE price_in_USD > 1000000
LIMIT 10;
```

```python
%%sql

SELECT 
    title, 
    location, 
    price_in_USD
FROM real_estate
WHERE country = 'Turkey'
    AND price_in_USD > 500000
LIMIT 10;
```

```python
%%sql

SELECT
    title, 
    country, 
    price_in_USD
FROM real_estate
WHERE country = 'Turkey'
    OR country = 'Spain'
LIMIT 10;
```

```python
%%sql

SELECT
    title, 
    country, 
    price_in_USD
FROM real_estate
WHERE country IN ('Turkey', 'Spain', 'Italy')
LIMIT 10;
```

```python
%%sql

SELECT 
    title, 
    country, 
    price_in_USD
FROM real_estate
WHERE price_in_USD BETWEEN 200000 AND 300000
LIMIT 10;
```

```python
%%sql

SELECT
    title, 
    country, 
    price_in_USD
FROM real_estate
ORDER BY price_in_USD DESC
LIMIT 10;
```

```python
%%sql

SELECT
    country,
    COUNT(*) AS property_count
FROM real_estate
GROUP BY country
ORDER BY property_count DESC
LIMIT 10;
```

```python
%%sql

SELECT
    country,
    AVG(price_in_USD) AS average_price
FROM real_estate
WHERE price_in_USD > 0
GROUP BY country
ORDER BY average_price DESC
LIMIT 10;
```

```python
%%sql

SELECT
    country,
    MIN(price_in_USD) AS min_price,
    MAX(price_in_USD) AS max_price
FROM real_estate
WHERE price_in_USD > 0
GROUP BY country
ORDER BY max_price DESC
LIMIT 10;
```

```python
%%sql

SELECT
    apartment_rooms,
    COUNT(*) AS property_count
FROM real_estate
WHERE apartment_rooms > 0
GROUP BY apartment_rooms
ORDER BY apartment_rooms;
```

```python
%%sql

SELECT
    country,
    AVG(price_in_USD) AS average_price
FROM real_estate
WHERE price_in_USD > 500000
GROUP BY country
HAVING COUNT(*) >= 100
ORDER BY average_price DESC
LIMIT 10;
```

```python
%%sql

SELECT
    country,
    COUNT(*) AS property_count,
    AVG(price_in_USD) AS average_price
FROM real_estate
WHERE price_in_USD > 0
GROUP BY country
HAVING COUNT(*) > 1000
ORDER BY property_count DESC
LIMIT 10;
```

```python
%%sql

SELECT
    country,
    apartment_rooms,
    COUNT(*) AS property_count
FROM real_estate
WHERE apartment_rooms > 0
GROUP BY country, apartment_rooms
ORDER BY country, apartment_rooms;
```

```python
%%sql

SELECT
    country, 
    apartment_rooms, 
    AVG(price_in_USD) AS average_price
FROM real_estate
WHERE apartment_rooms > 0
    AND price_in_USD > 0
GROUP BY country, apartment_rooms
ORDER BY average_price DESC
LIMIT 10;
```
