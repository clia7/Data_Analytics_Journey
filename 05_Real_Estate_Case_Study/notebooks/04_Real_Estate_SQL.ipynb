```python
# Import relevant libraries

import pandas as pd 
import sqlite3

# Import CSV
df = pd.read_csv('../data/world_real_estate_data(147k).csv')

# Create SQLite database and table
with sqlite3.connect('../data/real_estate.db') as conn:
    df.to_sql(
        name='real_estate', 
        con=conn, 
        index=False, 
        if_exists='replace'
)

print('Database successfully created and loaded')
```
