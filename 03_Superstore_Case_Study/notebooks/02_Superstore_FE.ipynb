```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Safe Data Loading with raw dataset

file_path = '../data/superstore.csv'

try:
    df = pd.read_csv(file_path, encoding='latin1')
    print('Load successfully')
except FileNotFoundError:
    print(f'File not found - check {file_path}')
except Exception as e:
    print(f'Unexpected Error - check {e}')
```

```python
# Inspection of Dataset
df.info()
```

**Note:** No missing values

```python
# Check Metrics
df.describe()
```

```python
# Check DF Header
df.head()
```

```python
# Check for duplicates
df.duplicated().sum()
```

**Note:** No Duplicates

```python
# Convert Order Date and Ship Date into pd.datetime
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%m/%d/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%m/%d/%Y')
```

```python
# Create new feature - Shipping Days
df['Shipping_Days'] = df['Ship Date'] - df['Order Date']

# Check for result
df['Shipping_Days'].value_counts()
```

```python
# Check Metrics of Shipping Days
df['Shipping_Days'].describe()
```

```python
# Convert Timedelta into days
df['Shipping_Days'] = df['Shipping_Days'].dt.days

# Check the result
df['Shipping_Days'].value_counts()
```

**Note:**
Supply chain efficiency and fulfillment speed are primary drivers of customer satisfaction and long-term retention. By calculating the `Shipping_Days` it is possible to transform raw logistical data into a measurable key performance indicator (KPI) for the supply chain.

```python
# Extracting Year, Month and DayofWeek from Order Date
df['Order_Year'] = df['Order Date'].dt.year
df['Order_Month'] = df['Order Date'].dt.month
df['Order_DayOfWeek'] = df['Order Date'].dt.dayofweek

# Check for result
df[['Order_Year', 'Order_Month', 'Order_DayOfWeek']].head()
```

**Note:** Transforming raw date strings into numerical components for capturing patterns.

```python
# Create new feature for Orders on Weekend
df['is_weekend_order'] = df['Order_DayOfWeek'].isin([5,6]).astype(int)

# Check the result
df['is_weekend_order'].value_counts()
```

**Note:** Purchasing behavior on weekends often diverges fundamentally from weekday transactions. This binary feature isolates the distinct customer personas and operational requirements.

```python
# Create new feature 'is_profitable' if profit is > 0
df['is_profitable'] = (df['Profit'] > 0).astype(int)

# Check for result
df['is_profitable'].value_counts()
```

**Note:** Margin maximization and cost control are fundamental to business viability.

```python
# Create new feature 'profit_margin' by dividing profit with sales
df['profit_margin'] = df['Profit'] / df['Sales']
```

**Note:** While absolute revenue (`Sales`) and total `Profit` reveal the scale of cash flowing through the business, they fail to measure true financial efficiency. The `profit_margin` standardizes profitability as a ratio.
