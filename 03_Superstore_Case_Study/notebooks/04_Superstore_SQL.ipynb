```python
# Import relevant libraries
import pandas as pd
import sqlite3

# Import CSV
df = pd.read_csv('../data/superstore.csv', encoding='windows-1252')

# Change columns name
df.columns = df.columns.str.replace(' ', '_').str.replace('-','_')

conn = sqlite3.connect('superstore.db')
df.to_sql('orders', conn, index=False, if_exists='replace')
conn.close()

print('Database successfully loaded and converted')
```

```python
%load_ext sql

import prettytable
prettytable.__dict__['DEFAULT'] = 10

%sql sqlite:///superstore.db
```

```python
%%sql
SELECT * FROM orders LIMIT 5;
```

```python
%%sql
SELECT Category, SUM(Sales) FROM orders GROUP BY Category;
```

```python
%%sql
SELECT Category,
    SUM(Sales) AS total_revenue,
    MAX(Sales) AS highest_price,
    MIN(Sales) AS lowest_price,
    ROUND(AVG(Sales), 2) AS avg_price
FROM orders 
GROUP BY Category
ORDER BY total_revenue DESC;
```

```python
%%sql
SELECT Sub_Category,
    SUM(Sales) AS total_revenue,
    MAX(Sales) AS highest_price,
    MIN(Sales) AS lowest_price,
    ROUND(AVG(Sales), 2) AS avg_price
FROM orders 
GROUP BY Sub_Category
ORDER BY total_revenue DESC;
```

```python
%%sql
SELECT Sub_Category,
    ROUND(SUM(Profit), 2) AS Profit
FROM orders
GROUP BY Sub_Category
ORDER BY Profit DESC;
```

```python
%%sql
SELECT City, 
    SUM(Sales) as total_rev
FROM orders
GROUP BY City
ORDER BY total_rev DESC
LIMIT 5;
```

```python
%%sql
SELECT Sub_Category, 
    SUM(Profit) as total_profit
FROM orders
WHERE Segment = 'Corporate'
GROUP BY Sub_Category
ORDER BY total_profit DESC
LIMIT 3;
```

```python
%%sql
SELECT Ship_Mode,
    ROUND(AVG(Discount), 2) as avg_discount
FROM orders
WHERE Category = 'Furniture'
GROUP BY Ship_Mode
ORDER BY avg_discount DESC;
```

```python
%%sql
SELECT Region,
    SUM(Sales) AS total_rev,
    SUM(Profit) AS total_profit
FROM orders
GROUP BY Region
ORDER BY total_profit ASC;
```
